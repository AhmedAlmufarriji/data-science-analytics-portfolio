## 1. استيراد المكتبات وقراءة البيانات
تُقرأ البيانات مباشرة باستخدام `sns.load_dataset("tips")`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
print("Dataset shape:", tips.shape)
print(tips.head())
```

## 2. Histogram — توزيع الإكراميات حسب اليوم

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.histplot(data=tips, x="tip", hue="day", element="step")
plt.title("Tip Distribution by Day")
plt.xlabel("Tip")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("01_histogram_tip_by_day.png", dpi=200)
plt.show()
```

## 3. Displot — عدد السجلات حسب اليوم والجنس

```python
sns.set_theme(style="whitegrid")
g = sns.displot(
    data=tips, x="day", hue="sex",
    hue_order=["Female", "Male"], multiple="dodge",
    shrink=0.8, palette="Set1", height=5, aspect=1.6
)
g.set_axis_labels("Day", "Count")
g.figure.suptitle("Number of Records by Day and Sex", y=1.03)
g.figure.savefig("02_displot_day_by_sex.png", dpi=200, bbox_inches="tight")
plt.show()
```

## 4. ECDF — التوزيع التراكمي للإكراميات حسب وقت الوجبة
ملاحظة: `summer` خريطة ألوان؛ لذلك تُمرر إلى `palette` لا إلى `color`.

```python
sns.set_theme(style="darkgrid", font_scale=1.25)
plt.figure(figsize=(10, 6))
sns.ecdfplot(data=tips, x="tip", hue="time", palette="summer", linewidth=3)
plt.title("ECDF of Tips by Meal Time")
plt.xlabel("Tip")
plt.ylabel("Proportion")
plt.tight_layout()
plt.savefig("03_ecdf_tip_by_time.png", dpi=200)
plt.show()
```
